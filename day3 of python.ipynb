```python
import numpy as np
import pandas as pd

# Student Names
names = ["Vijay", "Rahul", "Sneha", "Priya", "Arjun"]

# Generate random marks (5 students × 3 subjects)
# Python, AI, ML
np.random.seed(1)
marks = np.random.randint(60, 101, (5, 3))

# Create DataFrame
df = pd.DataFrame(marks, columns=["Python", "AI", "ML"])
df.insert(0, "Name", names)

# Calculate Average
df["Average"] = df[["Python", "AI", "ML"]].mean(axis=1)

# Result Column
df["Result"] = np.where(df["Average"] >= 75, "Pass", "Fail")

# Display Final DataFrame
print("Final DataFrame:\n")
print(df)

# Find Topper
topper = df.loc[df["Average"].idxmax()]

print("\nTopper")
print("Name :", topper["Name"])
print("Average :", round(topper["Average"], 2))
```
